```python
import os
import winreg  # Windows 레지스트리 접근용

def is_font_installed(font_name: str) -> bool:
    """
    시스템에 폰트가 설치되어 있는지 확인 (Windows 전용)
    """
    try:
        reg_path = r"SOFTWARE\Microsoft\Windows NT\CurrentVersion\Fonts"
        with winreg.OpenKey(winreg.HKEY_LOCAL_MACHINE, reg_path) as key:
            num_fonts = winreg.QueryInfoKey(key)[1]
            for i in range(num_fonts):
                font_entry = winreg.EnumValue(key, i)[0]
                if font_name.lower() in font_entry.lower():
                    return True
        return False
    except Exception as e:
        print(f"폰트 확인 중 오류: {e}")
        return False
```

```python
is_font_installed("NanumGothic")  # 예시로 Malgun Gothic 폰트 확인
```

```python
from fontTools.ttLib import TTFont
import matplotlib.font_manager as fm
import os

def get_detailed_font_names():
    font_paths = fm.findSystemFonts(fontpaths=None, fontext='ttf') + \
                 fm.findSystemFonts(fontpaths=None, fontext='otf')

    font_names = []
    for path in font_paths:
        try:
            font = TTFont(path)
            name = ""
            subfamily = ""
            for record in font['name'].names:
                if record.nameID == 1 and not name:
                    name = record.string.decode(errors='ignore')
                elif record.nameID == 2 and not subfamily:
                    subfamily = record.string.decode(errors='ignore')
            if name and subfamily and subfamily.lower() != "regular":
                full_name = f"{name} {subfamily}"
            else:
                full_name = name
            font_names.append(full_name.strip())
        except Exception:
            continue

    return sorted(set(font_names))

fonts = get_detailed_font_names()
for f in fonts:
    print(f)
```

```python
from fontTools.ttLib import TTFont
import matplotlib.font_manager as fm
import unicodedata

def get_multilingual_font_names():
    font_paths = fm.findSystemFonts(fontpaths=None, fontext='ttf') + \
                 fm.findSystemFonts(fontpaths=None, fontext='otf')

    font_dict = {}
    for path in font_paths:
        try:
            font = TTFont(path)
            records = font['name'].names
            names = {}
            for rec in records:
                try:
                    name_str = rec.string.decode(rec.getEncoding(), errors='ignore')
                    # Unicode 한글 포함 여부 확인
                    if any('HANGUL' in unicodedata.name(char, '') for char in name_str):
                        names[f"{rec.nameID}_{rec.langID}"] = name_str
                except:
                    continue
            if names:
                font_dict[path] = names
        except:
            continue
    return font_dict

fonts = get_multilingual_font_names()
for path, names in fonts.items():
    print(f"📁 {path}")
    for key, val in names.items():
        print(f"  {key}: {val}")
```

```python
from fontTools.ttLib import TTFont
import matplotlib.font_manager as fm
import unicodedata

def get_korean_font_names():
    font_paths = fm.findSystemFonts(fontpaths=None, fontext='ttf') + \
                 fm.findSystemFonts(fontpaths=None, fontext='otf')

    font_names = set()
    for path in font_paths:
        try:
            font = TTFont(path)
            for record in font['name'].names:
                if record.langID == 1042 and record.nameID in [1, 3, 4]:
                    try:
                        name = record.string.decode(record.getEncoding(), errors='ignore')
                        # 한글이 포함된 경우만 필터
                        if any('HANGUL' in unicodedata.name(ch, '') for ch in name):
                            font_names.add(name.strip())
                    except:
                        continue
        except:
            continue

    return sorted(font_names)

# 실행 예시
fonts = get_korean_font_names()
for font in fonts:
    print(font)
```

```python
# -*- coding: utf-8 -*-
"""
시스템 폰트 중 한글·영문 이름을 분리해 출력
pip install fonttools matplotlib
"""
from fontTools.ttLib import TTFont
import matplotlib.font_manager as fm
import unicodedata
import sys

def contains_hangul(txt: str) -> bool:
    """문자열에 한글 코드포인트가 포함되었는지 검사"""
    return any('HANGUL' in unicodedata.name(ch, '') for ch in txt)

def collect_font_names():
    # 시스템에 설치된 TTF·OTF 경로 확보
    paths = (
        fm.findSystemFonts(fontpaths=None, fontext='ttf') +
        fm.findSystemFonts(fontpaths=None, fontext='otf')
    )

    korean, english = set(), set()

    for p in paths:
        try:
            tt = TTFont(p)
            for rec in tt['name'].names:
                # 패밀리(1), 스타일(3), 풀네임(4)만 수집
                if rec.nameID not in (1, 3, 4):
                    continue
                # 디코딩(언어코드별 인코딩 자동 파악)
                try:
                    name = rec.string.decode(rec.getEncoding(), errors='ignore').strip()
                except Exception:
                    continue
                if not name:
                    continue
                # 한글 포함 여부로 분리
                (korean if contains_hangul(name) else english).add(name)
        except Exception:
            # 손상된 폰트 파일 등은 무시
            continue

    return sorted(korean), sorted(english)

if __name__ == "__main__":
    ko_fonts, en_fonts = collect_font_names()

    # --- 출력 예시 ---
    print("=== Korean Fonts ===")
    for f in ko_fonts:
        print(f)
    print("\n=== English Fonts ===")
    for f in en_fonts:
        print(f)

    print(f"\n총 {len(ko_fonts)}개(한글) + {len(en_fonts)}개(영어) = {len(ko_fonts)+len(en_fonts)}개")
```
