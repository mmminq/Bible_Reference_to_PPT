```python
import os

def absolute_path(file_path):
    # 절대경로가 이미 주어졌으면 그대로 반환, 아니면 현재 경로 기준으로 반환
    if os.path.isabs(file_path):
        return file_path
    else:
        return os.path.abspath(file_path)


# ------------------------- 성경 데이터 처리 -------------------------

def read_files_in_directory(directory):
    file_contents = []
    for filename in os.listdir(directory):
        if filename.endswith('.txt'):
            with open(os.path.join(directory, filename), 'r', encoding='utf-8') as file:
                content = file.read()
                file_contents.append(content)
    return file_contents

bible_books = [
    "창세기", "출애굽기", "레위기", "민수기", "신명기", "여호수아", "사사기", "룻기", "사무엘상", "사무엘하",
    "열왕기상", "열왕기하", "역대상", "역대하", "에스라", "느헤미야", "에스더", "욥기", "시편", "잠언",
    "전도서", "아가", "이사야", "예레미야", "예레미야애가", "에스겔", "다니엘", "호세아", "요엘", "아모스",
    "오바댜", "요나", "미가", "나훔", "하박국", "스바냐", "학개", "스가랴", "말라기", "마태복음", "마가복음",
    "누가복음", "요한복음", "사도행전", "로마서", "고린도전서", "고린도후서", "갈라디아서", "에베소서", "빌립보서", "골로새서",
    "데살로니가전서", "데살로니가후서", "디모데전서", "디모데후서", "디도서", "빌레몬서", "히브리서", "야고보서", "베드로전서", "베드로후서",
    "요한일서", "요한이서", "요한삼서", "유다서", "요한계시록"
]

texts = read_files_in_directory(absolute_path('C:/Users/kmk47/OneDrive/바탕 화면/참고구절_PPT_자동화/개역개정-text_new'))
bible_dict = {bible_books[i]: texts[i] for i in range(len(bible_books))}
```

```python
import re

def find_missing_verse_numbers(text):
    """
    주어진 성경 본문 텍스트에서 '장:절' 형태가 아닌 '장:문장' 형태(절 번호가 없는 경우)를 찾아 반환합니다.
    예: '고전14:33 하나님은...' (정상), '고전14:모든 성도가...' (이상)
    """
    # 패턴: (책이름)(장):(절 또는 문장) (내용)
    # 정상: (책이름)(장):(절번호) (내용)
    # 이상: (책이름)(장):(숫자가 아닌 문자) (내용)
    results = []
    # 줄 단위로 검사
    for line in text.split('\n'):
        m = re.match(r'([가-힣]+)(\d+):([^\s]+)', line)
        if m:
            verse = m.group(3)
            # 절 번호가 숫자가 아니면 이상 케이스로 간주
            if not verse.isdigit():
                results.append(line)
    return results

# 예시: bible_dict에서 모든 책을 검사
missing_verses = {}
for book, text in bible_dict.items():
    lines = find_missing_verse_numbers(text)
    if lines:
        missing_verses[book] = lines

# 결과 출력 예시
for book, lines in missing_verses.items():
    print(f"[{book}]")
    for line in lines:
        print(line)
```

```python
import re

def remove_angle_bracket_sections(text):
    """
    본문 내용 중에서 '<'로 시작해서 '>'로 끝나는 내용을 모두 감지하여 제거합니다.
    """
    return re.sub(r'<[^>]*> ', '', text)

# 리스트 texts의 각 항목에 적용
texts_no_brackets = [remove_angle_bracket_sections(t) for t in texts]
```
